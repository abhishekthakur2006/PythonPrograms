# Find Second Largest Number in a List

Assignment 3 solution and assert-based checks.

````python
def second_largest(numbers: list[float]) -> float:
    """Return the second-largest distinct number in a non-empty list."""
    distinct_numbers = sorted(set(numbers))
    if len(distinct_numbers) < 2:
        raise ValueError('at least two distinct numbers are required')
    return distinct_numbers[-2]
````

````python
assert second_largest([1, 3, 2]) == 2

assert second_largest([5, 5, 4]) == 4

assert second_largest([-1, -3, -2]) == -2

for values in ([], [7, 7]):
    try:
        second_largest(values)
    except ValueError:
        pass
    else:
        raise AssertionError('fewer than two distinct values should raise ValueError')
print("All test cases passed.")
````
