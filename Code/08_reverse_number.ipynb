# Reverse a Number

Assignment 3 solution and assert-based checks.

````python
def reverse_number(number: int) -> int:
    """Return the number with its digits reversed, preserving its sign."""
    reversed_digits = int(str(abs(number))[::-1])
    return -reversed_digits if number < 0 else reversed_digits
````

````python
assert reverse_number(1234) == 4321

assert reverse_number(-508) == -805

assert reverse_number(1200) == 21

assert reverse_number(0) == 0
print("All test cases passed.")
````
