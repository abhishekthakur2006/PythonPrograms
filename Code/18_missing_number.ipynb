# Find Missing Number in a List

Assignment 3 solution and assert-based checks.

````python
def find_missing_number(numbers: list[int]) -> int:
    """Find the missing value from a list containing distinct values from 0 to n."""
    count = len(numbers)
    if len(set(numbers)) != count or any((number < 0 or number > count for number in numbers)):
        raise ValueError('values must be distinct integers from 0 through the list length')
    return count * (count + 1) // 2 - sum(numbers)
````

````python
assert find_missing_number([3, 0, 1]) == 2

assert find_missing_number([0]) == 1

assert find_missing_number([]) == 0

for values in ([0, 0], [1, 4]):
    try:
        find_missing_number(values)
    except ValueError:
        pass
    else:
        raise AssertionError('invalid input should raise ValueError')
print("All test cases passed.")
````
