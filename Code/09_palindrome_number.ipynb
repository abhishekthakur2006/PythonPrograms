# Check Palindrome Number

Assignment 3 solution and assert-based checks.

````python
def is_palindrome_number(number: int) -> bool:
    """Return whether a non-negative integer reads the same backward."""
    if number < 0:
        return False
    return number == int(str(number)[::-1])
````

````python
assert is_palindrome_number(121) is True

assert is_palindrome_number(0) is True

assert is_palindrome_number(123) is False

assert is_palindrome_number(-121) is False
print("All test cases passed.")
````
