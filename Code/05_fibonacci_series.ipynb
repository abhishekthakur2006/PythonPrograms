# Generate Fibonacci Series

Assignment 3 solution and assert-based checks.

````python
def fibonacci_series(count: int) -> list[int]:
    """Return the first count terms of the Fibonacci sequence."""
    if count < 0:
        raise ValueError('count must be non-negative')
    sequence = []
    (first, second) = (0, 1)
    for _ in range(count):
        sequence.append(first)
        (first, second) = (second, first + second)
    return sequence
````

````python
assert fibonacci_series(0) == []

assert fibonacci_series(1) == [0]

assert fibonacci_series(7) == [0, 1, 1, 2, 3, 5, 8]

try:
    fibonacci_series(-1)
except ValueError:
    pass
else:
    raise AssertionError('negative count should raise ValueError')
print("All test cases passed.")
````
