# Print Prime Numbers in a Range

Assignment 3 solution and assert-based checks.

````python
def primes_in_range(start: int, end: int) -> list[int]:
    """Return all prime numbers in the inclusive range start to end."""
    primes = []
    for candidate in range(max(2, start), end + 1):
        is_candidate_prime = True
        divisor = 2
        while divisor * divisor <= candidate:
            if candidate % divisor == 0:
                is_candidate_prime = False
                break
            divisor += 1
        if is_candidate_prime:
            primes.append(candidate)
    return primes
````

````python
assert primes_in_range(1, 10) == [2, 3, 5, 7]

assert primes_in_range(14, 20) == [17, 19]

assert primes_in_range(-5, 2) == [2]

assert primes_in_range(10, 1) == []
print("All test cases passed.")
````
