# Check Palindrome String

Assignment 3 solution and assert-based checks.

````python
def is_palindrome_string(text: str) -> bool:
    """Check a string as a case-insensitive sequence of alphanumeric characters."""
    normalized = ''.join((character.casefold() for character in text if character.isalnum()))
    return normalized == ''.join(reversed(normalized))
````

````python
assert is_palindrome_string('Racecar') is True

assert is_palindrome_string('A man, a plan, a canal: Panama!') is True

assert is_palindrome_string('Python') is False

assert is_palindrome_string('!!!') is True
print("All test cases passed.")
````
