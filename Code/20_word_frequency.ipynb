# Word Frequency in a Sentence

Assignment 3 solution and assert-based checks.

````python
import re

def word_frequency(sentence: str) -> dict[str, int]:
    """Count case-insensitive words, ignoring punctuation."""
    frequencies = {}
    for word in re.findall('\\b\\w+\\b', sentence.casefold()):
        frequencies[word] = frequencies.get(word, 0) + 1
    return frequencies
````

````python
assert word_frequency('Hello, hello world!') == {'hello': 2, 'world': 1}

assert word_frequency('') == {}

assert word_frequency('One-two one.') == {'one': 2, 'two': 1}
print("All test cases passed.")
````
