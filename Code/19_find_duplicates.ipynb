# Find Duplicate Elements in a List

Assignment 3 solution and assert-based checks.

````python
def find_duplicates(numbers: list[int]) -> list[int]:
    """Return each duplicated value once, ordered by its second occurrence."""
    seen = set()
    duplicates = []
    reported = set()
    for number in numbers:
        if number in seen and number not in reported:
            duplicates.append(number)
            reported.add(number)
        seen.add(number)
    return duplicates
````

````python
assert find_duplicates([1, 2, 1, 3, 2, 1]) == [1, 2]

assert find_duplicates([]) == []

assert find_duplicates([4, 5, 6]) == []
print("All test cases passed.")
````
