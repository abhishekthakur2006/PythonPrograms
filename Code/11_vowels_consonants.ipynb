# Count Vowels and Consonants

Assignment 3 solution and assert-based checks.

````python
def count_vowels_consonants(text: str) -> tuple[int, int]:
    """Return the counts of vowels and consonants, ignoring other characters."""
    vowels = 0
    consonants = 0
    for character in text.casefold():
        if character.isalpha():
            if character in 'aeiou':
                vowels += 1
            else:
                consonants += 1
    return (vowels, consonants)
````

````python
assert count_vowels_consonants('Hello, World!') == (3, 7)

assert count_vowels_consonants('AEIOU') == (5, 0)

assert count_vowels_consonants('123 !') == (0, 0)
print("All test cases passed.")
````
