# Find Common Elements in Two Lists

Assignment 3 solution and assert-based checks.

````python
def common_elements(first: list[int], second: list[int]) -> list[int]:
    """Return unique common values in the order they first appear in first."""
    second_values = set(second)
    common = []
    seen = set()
    for value in first:
        if value in second_values and value not in seen:
            common.append(value)
            seen.add(value)
    return common
````

````python
assert common_elements([1, 2, 2, 3, 4], [2, 4, 4]) == [2, 4]

assert common_elements([1, 3], [2, 4]) == []

assert common_elements([], [1]) == []
print("All test cases passed.")
````
