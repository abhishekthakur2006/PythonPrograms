# Count Frequency of Characters

Assignment 3 solution and assert-based checks.

````python
def character_frequency(text: str) -> dict[str, int]:
    """Return the case-sensitive frequency of each character in text."""
    frequencies = {}
    for character in text:
        frequencies[character] = frequencies.get(character, 0) + 1
    return frequencies
````

````python
assert character_frequency('banana') == {'b': 1, 'a': 3, 'n': 2}

assert character_frequency('Aa!') == {'A': 1, 'a': 1, '!': 1}

assert character_frequency('') == {}
print("All test cases passed.")
````
